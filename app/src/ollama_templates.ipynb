```python
import ollama
```

Two types to request the model with ollama:
1. Chat: Messages structuce with role 'user' and 'assistant'. 實現一來一往的對話模式。
2. Generate: 語言模型基礎運作模式，文字接龍。給定一文本讓模型去生成接下來的文字。

```python
question_1 = "What is the capital of Taiwan?"
answer_1 = "The capital of Taiwan is Taipei."
question_2 = "Introduce the foods onto the last answer."
```

```python
# Chat
response = ollama.chat(
    model       = "llama3.2",
    messages    = [
        {"role": "user", "content": question_1},
         {"role": "assistant", "content": answer_1},
         {"role": "user", "content": question_2}
    ]
)
print(response['message']['content'])
```

```python
# Generate
prompt = f"""
<|start_header_id|>system<|end_header_id|><|eot_id|>
<|start_header_id|>user<|end_header_id|>
{question_1}
<|eot_id|>
<|start_header_id|>assistant<|end_header_id|>
{answer_1}
<|eot_id|>
<|start_header_id|>user<|end_header_id|>
{question_2}
<|eot_id|>
<|start_header_id|>assistant<|end_header_id|>
"""

response = ollama.generate(
    model       = "llama3.2",
    prompt      = prompt,
    stream      = False
)

print(response["response"])
```
